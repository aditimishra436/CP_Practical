```python
# get marks of students and give grades based on the marks
# 95 and above - grade A+, 90-94 - grade A, 80-89 - grade B, 75-79 - grade C, 70-74 - grade D, below 70 - grade F
marks = float(input('What are your marks? '))
if marks >= 95:
    print('You got grade A+')
elif marks >= 90:
    print('You got grade A')
elif marks >= 80:
    print('You got grade B')
elif marks >= 75:
    print('You got grade C')
elif marks >= 70:
    print('You got grade D')
else:
    print('You got grade F')
```
